```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.interpolate import make_interp_spline
import io

# Set modern Seaborn style
sns.set_theme(style="whitegrid", palette="muted")
sns.set_context("notebook", font_scale=1.1)

df = pd.read_csv(r'F:\Fraunhofer_Backup\2026_09_14\mnt\checkpoints\3d\manhole_seg_pointnet2_first_try_run6\seg_history_PointNet2_MSG_Seg.csv')
```

```python
def smooth_line(x, y):
    """Interpolates points to create a smooth curve."""
    x_smooth = np.linspace(x.min(), x.max(), 300)
    spline = make_interp_spline(x, y, k=3) 
    y_smooth = spline(x_smooth)
    return x_smooth, y_smooth

plt.figure(figsize=(10, 6))

# Generate smooth curves
x_train, y_train = smooth_line(df['epoch'], df['train_loss'])
x_val, y_val = smooth_line(df['epoch'], df['val_loss'])

# Plot smooth lines
plt.plot(x_train, y_train, label='Training Loss', color='#3498db', linewidth=1)
plt.plot(x_val, y_val, label='Validation Loss', color='#e74c3c', linewidth=1)

# Overlay original data points as dots
plt.scatter(df['epoch'], df['train_loss'], color='#2980b9', s=20, zorder=3, alpha=0.7)
plt.scatter(df['epoch'], df['val_loss'], color='#c0392b', s=20, zorder=3, alpha=0.7)

# Modernize the axes and labels
plt.title('Loss Over Epochs', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Epoch', fontsize=12)
plt.ylabel('Loss', fontsize=12)
plt.legend(frameon=True, fancybox=True, shadow=True, borderpad=1)

# Remove top and right spines for a cleaner look
sns.despine()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df['epoch'], df['train_loss'], label='Training Loss', marker='o', linewidth=1)
plt.plot(df['epoch'], df['val_loss'], label='Validation Loss', marker='s', linewidth=1)

plt.title('Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df['epoch'], df['train_acc'], label='Train Accuracy', marker='o', linewidth=1)
plt.plot(df['epoch'], df['val_acc'], label='Val Accuracy', marker='s', linewidth=1)

plt.title('Accuracy Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy (%)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
